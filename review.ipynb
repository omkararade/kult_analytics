```python
import pandas as pd
from google_play_scraper import Sort, reviews
import time
import logging

# Set up logging to monitor progress
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

app_id = 'mate.bluetoothprint'  # Replace with your app’s ID
csv_file = 'reviews_bluetooth.csv'
batch_size = 200  # Number of reviews per request

# Save reviews to CSV
def save_to_csv(reviews_list):
    if reviews_list:
        df = pd.DataFrame(reviews_list)
        df.to_csv(csv_file, mode='a', header=not pd.io.common.file_exists(csv_file), index=False)
        logging.info(f"Saved {len(reviews_list)} reviews")

# Fetch all reviews
continuation_token = None
fetched_reviews = 0
request_count = 0
max_requests = 150  # Safety limit

while request_count < max_requests:
    try:
        result, continuation_token = reviews(
            app_id,  # Adjust if you need other languages
            sort=Sort.NEWEST,
            count=batch_size,
            continuation_token=continuation_token
        )
        if result:
            save_to_csv(result)
            fetched_reviews += len(result)
            request_count += 1
            logging.info(f"Fetched {len(result)} reviews, Total: {fetched_reviews}")
            time.sleep(5)  # Delay to avoid rate limits
        else:
            logging.info("No more reviews to fetch")
            break
    except Exception as e:
        logging.error(f"Error: {e}")
        time.sleep(10)  # Wait longer if there’s an error

    if continuation_token is None:
        logging.info("Reached the end of reviews")
        break

# Remove duplicates
df = pd.read_csv(csv_file)
df = df.drop_duplicates(subset='reviewId')
df.to_csv(csv_file, index=False)
logging.info(f"Final unique reviews: {len(df)}")
```
